```python
import sys
```

```python
var_str = 'string'
var_int = 23
var_bool = True

var_list = []
var_tuple = ()
var_dict = {}

# primitive type
print(sys.getrefcount(var_str))
print(sys.getrefcount(var_int))
print(sys.getrefcount(var_bool))

# collection types
print(sys.getrefcount(var_list))
print(sys.getrefcount(var_tuple))
print(sys.getrefcount(var_dict))
```
